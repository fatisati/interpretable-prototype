```python
import pandas as pd
import numpy as np

def show(df):
    def style_col(col):
        vals = col.values.astype(float)
        order = np.argsort(-vals)  # descending
        styles = [""] * len(col)
        if len(order) > 0:
            styles[order[0]] = "font-weight: bold"
        if len(order) > 1:
            styles[order[1]] = "text-decoration: underline"
        return styles

    return df.style.apply(style_col, axis=0)
```
